BINARY TREE

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


# Membuat binary tree secara manual
root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)


# Fungsi inorder traversal
def inorder(node):
    if node:
        inorder(node.left)
        print(node.data, end=' ')
        inorder(node.right)


# Menampilkan hasil traversal
print("Hasil inorder traversal dari binary tree:")
inorder(root)
```

BINARY SEARCH TREE

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


class BinarySearchTree:
    def __init__(self):
        self.root = None

    def insert(self, root, data):
        # Jika node kosong, buat node baru
        if root is None:
            return Node(data)

        # Masukkan ke subtree kiri jika data lebih kecil
        if data < root.data:
            root.left = self.insert(root.left, data)
        else:
            # Masukkan ke subtree kanan jika data lebih besar atau sama
            root.right = self.insert(root.right, data)

        return root


# Membuat Binary Search Tree
bst = BinarySearchTree()
root = None

# Data yang akan dimasukkan
data_list = [50, 30, 70, 20, 40, 60, 80]

# Proses insert data ke BST
for value in data_list:
    root = bst.insert(root, value)
```

TRAVERSAL PADA BTS

```python
# Fungsi inorder traversal
def inorder(node):
    if node:
        inorder(node.left)
        print(node.data, end=' ')
        inorder(node.right)


# Fungsi preorder traversal
def preorder(node):
    if node:
        print(node.data, end=' ')
        preorder(node.left)
        preorder(node.right)


# Fungsi postorder traversal
def postorder(node):
    if node:
        postorder(node.left)
        postorder(node.right)
        print(node.data, end=' ')


# Menampilkan hasil traversal
print("Inorder Traversal:")
inorder(root)

print("\nPreorder Traversal:")
preorder(root)

print("\nPostorder Traversal:")
postorder(root)
```

```python
# Program Binary Search Tree (BST)

# -----------------------------------
# Class Node untuk membuat node BST
# -----------------------------------
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


# -----------------------------------
# Class Binary Search Tree
# -----------------------------------
class BinarySearchTree:
    def __init__(self):
        self.root = None

    # Fungsi insert data ke BST
    def insert(self, root, data):
        if root is None:
            return Node(data)

        if data < root.data:
            root.left = self.insert(root.left, data)
        else:
            root.right = self.insert(root.right, data)

        return root


# -----------------------------------
# Fungsi Traversal
# -----------------------------------

# Inorder Traversal
def inorder(node):
    if node:
        inorder(node.left)
        print(node.data, end=' ')
        inorder(node.right)


# Preorder Traversal
def preorder(node):
    if node:
        print(node.data, end=' ')
        preorder(node.left)
        preorder(node.right)


# Postorder Traversal
def postorder(node):
    if node:
        postorder(node.left)
        postorder(node.right)
        print(node.data, end=' ')


# -----------------------------------
# Fungsi Pencarian pada BST
# -----------------------------------
def search(node, key):
    # Jika node kosong atau data ditemukan
    if node is None or node.data == key:
        return node

    # Cari ke subtree kiri
    if key < node.data:
        return search(node.left, key)

    # Cari ke subtree kanan
    return search(node.right, key)


# -----------------------------------
# Data identitas
# -----------------------------------

# Contoh data
nama_depan = "Adnan"
npm = "2505060064"
tanggal_lahir = "21-09-2006"

# 1. Tiga digit terakhir NPM
tiga_digit_npm = int(npm[-3:])

# 2. Tanggal lahir dipisah menjadi DD dan MM
tanggal = int(tanggal_lahir.split("-")[0])
bulan = int(tanggal_lahir.split("-")[1])

# 3. ASCII huruf pertama dan kedua nama depan
ascii_1 = ord(nama_depan[0])  # A
ascii_2 = ord(nama_depan[1])  # D

# 4. Dua angka favorit
angka_favorit_1 = 67
angka_favorit_2 = 76

# Membuat list data BST
data_list = [
    tiga_digit_npm,
    tanggal,
    bulan,
    ascii_1,
    ascii_2,
    angka_favorit_1,
    angka_favorit_2
]


# -----------------------------------
# Membentuk BST
# -----------------------------------
bst = BinarySearchTree()
root = None

for data in data_list:
    root = bst.insert(root, data)


# -----------------------------------
# Menampilkan hasil traversal
# -----------------------------------
print("Inorder Traversal:")
inorder(root)

print("\n\nPreorder Traversal:")
preorder(root)

print("\n\nPostorder Traversal:")
postorder(root)


# -----------------------------------
# Proses pencarian data
# -----------------------------------

# Mencari dua digit terakhir NPM
cari_1 = int(npm[-2:])

# Mencari angka yang tidak ada dalam list
cari_2 = 1000

print("\n\nHasil Pencarian:")

# Pencarian pertama
hasil = search(root, cari_1)
if hasil:
    print(f"{cari_1} ditemukan dalam BST.")
else:
    print(f"{cari_1} tidak ditemukan dalam BST.")

# Pencarian kedua
hasil = search(root, cari_2)
if hasil:
    print(f"{cari_2} ditemukan dalam BST.")
else:
    print(f"{cari_2} tidak ditemukan dalam BST.")
```
