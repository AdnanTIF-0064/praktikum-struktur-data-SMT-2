GRAF DENGAN ADJACENCY LIST

```python
# Membuat graf tak berarah menggunakan adjacency list
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}


# Menampilkan graf
def print_graph(graph):
    for node in graph:
        print(f"{node} -> {graph[node]}")


# Memanggil fungsi print_graph
print_graph(graph)
```

TRAVERSAL GRAF DENGAN BFS

```python
from collections import deque


# Fungsi BFS (Breadth First Search)
def bfs(graph, start):
    visited = set()
    queue = deque([start])

    while queue:
        vertex = queue.popleft()

        if vertex not in visited:
            print(vertex, end=" ")
            visited.add(vertex)

            # Menambahkan tetangga yang belum dikunjungi ke queue
            queue.extend(
                neighbor
                for neighbor in graph[vertex]
                if neighbor not in visited
            )


# Contoh pemanggilan BFS
bfs(graph, 'A')
```

TRAVERSAL GRAF DENGAN DFS

```python
def dfs(graph, start, visited=None):
    if visited is None:
        visited = set()

    if start not in visited:
        print(start, end=" ")
        visited.add(start)
        for neighbor in graph[start]:
            dfs(graph, neighbor, visited)

# Contoh pemanggilan
dfs(graph, 'A')
```

TUGAS

```python
# Program Graph dengan BFS, DFS, Find Path, dan Cek Koneksi

from collections import deque


# -----------------------------------
# Class Graph
# -----------------------------------
class Graph:
    def __init__(self):
        # Dictionary untuk menyimpan adjacency list
        self.graph = {}

    # Menambahkan sisi (edge) tak berarah
    def add_edge(self, u, v):
        if u not in self.graph:
            self.graph[u] = []

        if v not in self.graph:
            self.graph[v] = []

        self.graph[u].append(v)
        self.graph[v].append(u)

    # Menampilkan graph
    def print_graph(self):
        print("Adjacency List Graph:")
        for node in self.graph:
            print(f"{node} -> {self.graph[node]}")

    # -----------------------------------
    # BFS (Mengembalikan list kunjungan)
    # -----------------------------------
    def bfs(self, start):
        visited = set()
        queue = deque([start])

        hasil = []

        visited.add(start)

        while queue:
            node = queue.popleft()
            hasil.append(node)

            for neighbor in self.graph[node]:
                if neighbor not in visited:
                    visited.add(neighbor)
                    queue.append(neighbor)

        return hasil

    # -----------------------------------
    # DFS
    # -----------------------------------
    def dfs(self, start, visited=None):
        if visited is None:
            visited = set()

        visited.add(start)
        print(start, end=' ')

        for neighbor in self.graph[start]:
            if neighbor not in visited:
                self.dfs(neighbor, visited)

    # -----------------------------------
    # Find Path menggunakan DFS
    # -----------------------------------
    def find_path(self, start, end, path=None):
        if path is None:
            path = []

        path = path + [start]

        # Jika tujuan ditemukan
        if start == end:
            return path

        # Jika node tidak ada
        if start not in self.graph:
            return None

        for neighbor in self.graph[start]:
            if neighbor not in path:
                new_path = self.find_path(neighbor, end, path)

                if new_path:
                    return new_path

        return None

    # -----------------------------------
    # Mengecek apakah graph terhubung
    # -----------------------------------
    def is_connected(self):
        # Ambil salah satu simpul awal
        start_node = next(iter(self.graph))

        visited = set()

        # DFS untuk menandai node yang dikunjungi
        self._dfs_connected(start_node, visited)

        # Jika semua node dikunjungi
        return len(visited) == len(self.graph)

    # Fungsi DFS bantuan untuk konektivitas
    def _dfs_connected(self, node, visited):
        visited.add(node)

        for neighbor in self.graph[node]:
            if neighbor not in visited:
                self._dfs_connected(neighbor, visited)


# -----------------------------------
# Membuat Graph Baru
# -----------------------------------
g = Graph()

# Menambahkan minimal 7 sisi
g.add_edge('A', 'B')
g.add_edge('A', 'C')
g.add_edge('B', 'D')
g.add_edge('C', 'D')
g.add_edge('C', 'E')
g.add_edge('D', 'E')
g.add_edge('B', 'E')


# -----------------------------------
# Menampilkan Graph
# -----------------------------------
g.print_graph()


# -----------------------------------
# BFS
# -----------------------------------
print("\nBFS Traversal dari A:")
hasil_bfs = g.bfs('A')
print(hasil_bfs)


# -----------------------------------
# DFS
# -----------------------------------
print("\nDFS Traversal dari A:")
g.dfs('A')


# -----------------------------------
# Find Path
# -----------------------------------
print("\n\nPath dari A ke E:")
path = g.find_path('A', 'E')

if path:
    print(path)
else:
    print("Path tidak ditemukan")


# -----------------------------------
# Cek Koneksi Graph
# -----------------------------------
print("\nApakah graph terhubung?")

if g.is_connected():
    print("Graph terhubung")
else:
    print("Graph tidak terhubung")
```
